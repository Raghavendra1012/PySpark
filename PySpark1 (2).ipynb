```python
from pyspark import SparkContext

sc = SparkContext ("local", "RDD_revision")
```

```python
list1 = (1,2,3,4,5,6,7,8)

print(type(list1))

rdd1 = sc.parallelize(list1)

print(type(rdd1))

rdd2 = rdd1.filter( lambda x: x>=5)
rdd2.collect()
```

```python
list1 = (1,2,3,4,5,6,7,8)

print(type(list1))

rdd1 = sc.parallelize(list1)

print(type(rdd1))

rdd2 = rdd1.map (lambda y: y+100)

rdd2.collect()
```
